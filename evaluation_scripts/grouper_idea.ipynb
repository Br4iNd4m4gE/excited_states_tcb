```python
import os
import numpy as np
import itertools
```

```python
def grouper(iterable, n, fillvalue = None):
    """
    Reads chunks of n lines from iterable. Fills missing lines at the end with fillvalue.
    Taken from the recipes in the itertools docs.
    """
    args = [iter(iterable)] * n
    return itertools.zip_longest(*args, fillvalue=fillvalue)
```

```python
lines_per_geom = 2

x = [1,2,3,4,5]
# erg = grouper(x, lines_per_geom)
args = [iter(x)] * lines_per_geom
print(args)
print(len(args))
erg = itertools.zip_longest(*args, fillvalue=None)
print(erg)
for i in erg:
    print(i)
    print('')
```

```python
x = [[1,1,6,4,1,3], [2,3,2,3,4,5]]
z = itertools.zip_longest(*x, fillvalue=None)
print(z)
for i in z:
    print(i)
    print('')
```
