```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import binom
```

```python
# Define bionmial distribution
def binomial(n, k_0, x, p):
    return np.sum([binom(n, k) * p**k * (1-p)**(n-k) for k in range(int(k_0), int(x+1))])
```

```python
p = 0.04
n = 800
x = 29
k_0 = 28

binomial(n, k_0, x, p)
```
