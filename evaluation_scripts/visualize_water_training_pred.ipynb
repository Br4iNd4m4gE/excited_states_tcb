```python
import numpy as np
import matplotlib.pyplot as plt
from os.path import join, isfile, isdir
```

```python
all_solvent_names = [
    'acn',
    'meoh',
    'dcm',
    'dmso',
    'acetone',
    'water',
    'dmf',
    'hexane',
    'ethylacetate',
    'dioxane'
    ]
```

```python
parent_path = '/data/cschmidt/excited_states/energy_NNs/fr0_project'

model_prefix = 'train_only_wrong_'

best_model = 'best_model'

energy_eV_ref_file = 'hp_ref_energies_eV.dat'
energy_eV_pred_file = 'hp_predicted_energies_eV.dat'

osc_ref_file = 'hp_ref_osc.dat'
osc_pred_file = 'hp_predicted_osc.dat'

model_paths = [join(parent_path, f'model_prefix{solvent}', best_model) for solvent in all_solvent_names if isdir(join(parent_path, f'model_prefix{solvent}', best_model))]

energy_eV_ref_path = join(parent_path, energy_eV_ref_file)
energy_eV_pred_path = join(parent_path, energy_eV_pred_file)

osc_ref_path = join(parent_path, osc_ref_file)
osc_pred_path = join(parent_path, osc_pred_file)
```

```python
energy_eV_ref = np.loadtxt(energy_eV_ref_path)
energy_eV_pred = np.loadtxt(energy_eV_pred_path)

osc_ref = np.loadtxt(osc_ref_path)
osc_pred = np.loadtxt(osc_pred_path)

print(len(energy_eV_ref), len(energy_eV_pred), len(osc_ref), len(osc_pred))
```

```python
plt.plot(energy_eV_ref, energy_eV_pred, 'o')
plt.xlabel('Reference energy [eV]')
plt.ylabel('Predicted energy [eV]')
plt.show()

plt.plot(osc_ref, osc_pred, 'o')
plt.xlabel('Reference oscillator strength')
plt.ylabel('Predicted oscillator strength')
plt.show()
```
