```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow.keras as ks
```

```python
# Define the leaky_softplus function
def leaky_softplus(alpha=0.3):
    """
    Leaky softplus activation function similar to leakyRELU but smooth.
        
    Args:
        alpha (float, optional): Leaking slope. The default is 0.3.

    Returns:
        func: lambda function of x.

    """
    return lambda x: ks.activations.softplus(x) * (1 - alpha) + alpha * x

# Generate sample data
x_values = np.linspace(-10, 10, 400)

# Get the leaky_softplus function with a specific alpha
alpha = 0.3
activation_function = leaky_softplus(alpha)

# Apply the activation function to the sample data
y_values = activation_function(x_values)
y_softplus = ks.activations.softplus(x_values)

# Plot the function
plt.figure(figsize=(8, 6))
plt.plot(x_values, y_values, label=f'leaky_softplus(alpha={alpha})')
plt.plot(x_values, y_softplus, label='softplus')
plt.title('Leaky Softplus Activation Function')
plt.xlabel('Input')
plt.ylabel('Output')
plt.legend()
plt.grid(True)
plt.show()
```
